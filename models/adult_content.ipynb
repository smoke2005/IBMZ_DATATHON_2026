```python
# ============================================================
# ADULT / NON-ADULT TEXT CLASSIFICATION EXPERIMENT PIPELINE
# DistilBERT randomized hyperparameter search + classical ML baselines
# Validation-based model selection + ROC/PR curves + visualizations
# Run this entire script in ONE Jupyter cell.
# ============================================================

import re, gc, json, random, warnings, inspect
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import torch

from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, roc_curve, roc_auc_score,
    precision_recall_curve, average_precision_score
)
from datasets import Dataset
from transformers import (
    AutoTokenizer, AutoModelForSequenceClassification,
    TrainingArguments, Trainer, DataCollatorWithPadding
)

warnings.filterwarnings("ignore", category=FutureWarning)


# Compatibility helpers for different Transformers versions.
# Older releases use evaluation_strategy and tokenizer=; newer releases
# may use eval_strategy and processing_class=. Unsupported optional arguments
# are filtered out rather than crashing the notebook.
def make_training_args(**kwargs):
    supported = inspect.signature(TrainingArguments.__init__).parameters
    if "eval_strategy" in kwargs and "eval_strategy" not in supported and "evaluation_strategy" in supported:
        kwargs["evaluation_strategy"] = kwargs.pop("eval_strategy")
    # warmup_ratio is not present in some older Transformers releases.
    # In that case, omit it; the remaining training settings still work.
    kwargs = {key: value for key, value in kwargs.items() if key in supported}
    return TrainingArguments(**kwargs)

def make_trainer(**kwargs):
    supported = inspect.signature(Trainer.__init__).parameters
    if "processing_class" in kwargs and "processing_class" not in supported:
        kwargs["tokenizer"] = kwargs.pop("processing_class")
    kwargs = {key: value for key, value in kwargs.items() if key in supported}
    return Trainer(**kwargs)

# ============================================================
# 1. PATHS AND CONFIGURATION
# ============================================================
BASE_DIR = Path.cwd()
PROJECT_DIR = BASE_DIR.parent  # notebook expected to be inside scripts/
DATASET_PATH = PROJECT_DIR / "data" / "Copy of final_adult_content(1).csv"
OUTPUT_DIR = PROJECT_DIR / "model_experiments"
TRIALS_DIR = OUTPUT_DIR / "distilbert_trials"
MODEL_PATH = PROJECT_DIR / "models" / "adult_distilbert"
MODEL_NAME = "distilbert/distilbert-base-uncased"
MAX_LENGTH = 256
SEED = 42

# Start with a small search, especially on CPU. Add configurations later if needed.
SEARCH_CONFIGS = [
    {"learning_rate": 2e-5, "epochs": 2, "batch_size": 8, "weight_decay": 0.01, "warmup_ratio": 0.0},
    {"learning_rate": 3e-5, "epochs": 2, "batch_size": 8, "weight_decay": 0.01, "warmup_ratio": 0.1},
    {"learning_rate": 2e-5, "epochs": 3, "batch_size": 8, "weight_decay": 0.05, "warmup_ratio": 0.1},
    {"learning_rate": 5e-5, "epochs": 2, "batch_size": 8, "weight_decay": 0.01, "warmup_ratio": 0.1},
]
LABEL2ID = {"non-adult": 0, "adult": 1}
ID2LABEL = {0: "non-adult", 1: "adult"}
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
TRIALS_DIR.mkdir(parents=True, exist_ok=True)
MODEL_PATH.mkdir(parents=True, exist_ok=True)

print("Project directory:", PROJECT_DIR)
print("Dataset path:", DATASET_PATH)
print("Dataset exists:", DATASET_PATH.exists())
print("PyTorch:", torch.__version__, "| CUDA available:", torch.cuda.is_available())
if not DATASET_PATH.exists():
    print("CSV files found under project directory:")
    for p in PROJECT_DIR.rglob("*.csv"):
        print(" -", p)
    raise FileNotFoundError(f"Dataset not found: {DATASET_PATH}. Update DATASET_PATH above.")

# ============================================================
# 2. REPRODUCIBILITY AND PREPROCESSING
# ============================================================
def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
set_seed()

def normalize_label(value):
    if pd.isna(value):
        return None
    label = re.sub(r"[\s_]+", "-", str(value).strip().lower())
    return {
        "adult": "adult", "unsafe": "adult",
        "non-adult": "non-adult", "nonadult": "non-adult", "safe": "non-adult"
    }.get(label)

def clean_text(value):
    return re.sub(r"\s+", " ", str(value)).strip()

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip().str.lower()
if not {"description", "category"}.issubset(df.columns):
    raise ValueError(f"CSV needs Description and Category columns. Found: {df.columns.tolist()}")

df = df[["description", "category"]].dropna().copy()
df["text"] = df["description"].apply(clean_text)
df["category"] = df["category"].apply(normalize_label)
df = df.dropna(subset=["category"])
df = df[df["text"] != ""].copy()
df["label"] = df["category"].map(LABEL2ID)
df = df.drop_duplicates(subset=["text", "label"])
conflicts = df.groupby("text")["label"].nunique()
conflicts = conflicts[conflicts > 1].index
if len(conflicts):
    print("Conflicting descriptions removed:", len(conflicts))
    df = df[~df["text"].isin(conflicts)].copy()
df = df[["text", "label"]].reset_index(drop=True)
if df.empty or df["label"].nunique() != 2:
    raise ValueError("Both adult and non-adult classes must remain after cleaning.")
if df["label"].value_counts().min() < 4:
    raise ValueError("Need at least four examples in each class for these splits.")

print("\nCleaned dataset shape:", df.shape)
print("Class counts:\n", df["label"].map(ID2LABEL).value_counts())

# Train 65%, validation 15%, test 20%. Test is not used for tuning/model selection.
train_val_df, test_df = train_test_split(
    df, test_size=0.20, random_state=SEED, stratify=df["label"]
)
train_df, val_df = train_test_split(
    train_val_df, test_size=0.1875, random_state=SEED, stratify=train_val_df["label"]
)
train_df, val_df, test_df = [x.reset_index(drop=True) for x in (train_df, val_df, test_df)]
X_train, y_train = train_df["text"].tolist(), train_df["label"].to_numpy()
X_val, y_val = val_df["text"].tolist(), val_df["label"].to_numpy()
X_test, y_test = test_df["text"].tolist(), test_df["label"].to_numpy()
print(f"\nSplit sizes: train={len(train_df)}, validation={len(val_df)}, test={len(test_df)}")

# ============================================================
# 3. SHARED METRICS
# ============================================================
def get_metrics(y_true, y_pred, scores=None):
    m = {
        "accuracy": accuracy_score(y_true, y_pred),
        "adult_precision": precision_score(y_true, y_pred, pos_label=1, zero_division=0),
        "adult_recall": recall_score(y_true, y_pred, pos_label=1, zero_division=0),
        "adult_f1": f1_score(y_true, y_pred, pos_label=1, zero_division=0),
    }
    if scores is not None and len(np.unique(y_true)) == 2:
        m["roc_auc"] = roc_auc_score(y_true, scores)
        m["pr_auc"] = average_precision_score(y_true, scores)
    return m

def hf_metrics(eval_pred):
    logits, labels = eval_pred
    pred = np.argmax(logits, axis=-1)
    return {
        "accuracy": accuracy_score(labels, pred),
        "adult_precision": precision_score(labels, pred, pos_label=1, zero_division=0),
        "adult_recall": recall_score(labels, pred, pos_label=1, zero_division=0),
        "adult_f1": f1_score(labels, pred, pos_label=1, zero_division=0),
    }

# ============================================================
# 4. CLASSICAL ML BASELINES: TF-IDF + LOGISTIC REGRESSION / LINEAR SVM
# ============================================================
classical_models = {
    "TF-IDF + Logistic Regression": Pipeline([
        ("tfidf", TfidfVectorizer(max_features=100000, ngram_range=(1, 2), sublinear_tf=True)),
        ("classifier", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=SEED)),
    ]),
    "TF-IDF + Linear SVM": Pipeline([
        ("tfidf", TfidfVectorizer(max_features=100000, ngram_range=(1, 2), sublinear_tf=True)),
        ("classifier", LinearSVC(class_weight="balanced", random_state=SEED)),
    ]),
}
classical_validation = {}
classical_test_outputs = {}
classical_fitted = {}
classical_cv_results = {}
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=SEED)
search_spaces = {
    "TF-IDF + Logistic Regression": {
        "tfidf__ngram_range": [(1, 1), (1, 2)],
        "tfidf__max_features": [50000, 100000],
        "classifier__C": [0.1, 1.0, 3.0, 10.0],
    },
    "TF-IDF + Linear SVM": {
        "tfidf__ngram_range": [(1, 1), (1, 2)],
        "tfidf__max_features": [50000, 100000],
        "classifier__C": [0.1, 0.5, 1.0, 2.0, 5.0],
    },
}
print("\n" + "=" * 60 + "\nCLASSICAL ML: RANDOM SEARCH WITH 3-FOLD STRATIFIED CV\n" + "=" * 60)
for name, pipeline in classical_models.items():
    print("\nTuning:", name)
    search = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions=search_spaces[name],
        n_iter=4,
        scoring="f1",
        cv=cv,
        refit=True,
        random_state=SEED,
        n_jobs=1,
        verbose=1,
        return_train_score=True,
    )
    search.fit(X_train, y_train)
    fitted_pipeline = search.best_estimator_
    classical_cv_results[name] = {
        "best_cv_f1": float(search.best_score_),
        "best_params": search.best_params_,
    }
    print("Best 3-fold CV F1:", search.best_score_)
    print("Best parameters:", search.best_params_)

    val_pred = fitted_pipeline.predict(X_val)
    val_score = fitted_pipeline.predict_proba(X_val)[:, 1] if hasattr(fitted_pipeline, "predict_proba") else fitted_pipeline.decision_function(X_val)
    classical_validation[name] = get_metrics(y_val, val_pred, val_score)
    classical_fitted[name] = fitted_pipeline
    test_pred = fitted_pipeline.predict(X_test)
    test_score = fitted_pipeline.predict_proba(X_test)[:, 1] if hasattr(fitted_pipeline, "predict_proba") else fitted_pipeline.decision_function(X_test)
    classical_test_outputs[name] = {"y_pred": np.asarray(test_pred), "y_score": np.asarray(test_score), "metrics": get_metrics(y_test, test_pred, test_score)}
    print("Validation metrics:", classical_validation[name])

pd.DataFrame(classical_cv_results).T.to_csv(OUTPUT_DIR / "classical_ml_cv_search_results.csv")

# ============================================================
# 5. PREPARE DISTILBERT DATASETS
# ============================================================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

def make_hf_dataset(frame):
    hf_frame = frame[["text", "label"]].rename(columns={"label": "labels"}).copy()
    ds = Dataset.from_pandas(hf_frame.reset_index(drop=True))
    def tokenize_batch(batch):
        return tokenizer(batch["text"], truncation=True, max_length=MAX_LENGTH)
    return ds.map(tokenize_batch, batched=True, remove_columns=["text"])

train_ds = make_hf_dataset(train_df)
val_ds = make_hf_dataset(val_df)
test_ds = make_hf_dataset(test_df)

# ============================================================
# 6. RANDOM SEARCH FOR DISTILBERT
#    The validation adult-F1 chooses the best configuration.
# ============================================================
trial_records = []
best_val_f1 = -1.0
best_trial_model_dir = None
best_config = None
print("\n" + "=" * 60 + "\nDISTILBERT RANDOM SEARCH\n" + "=" * 60)
print("Number of configurations:", len(SEARCH_CONFIGS))
print("CPU training may take a while; each configuration trains separately.")

for trial, config in enumerate(SEARCH_CONFIGS, start=1):
    trial_dir = TRIALS_DIR / f"trial_{trial}"
    checkpoint_dir = trial_dir / "checkpoints"
    best_dir = trial_dir / "best_model"
    trial_dir.mkdir(parents=True, exist_ok=True)
    print(f"\n--- Trial {trial}/{len(SEARCH_CONFIGS)} ---\n{config}")

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME, num_labels=2, label2id=LABEL2ID, id2label=ID2LABEL
    )
    args = make_training_args(
        output_dir=str(checkpoint_dir),
        num_train_epochs=config["epochs"],
        learning_rate=config["learning_rate"],
        per_device_train_batch_size=config["batch_size"],
        per_device_eval_batch_size=config["batch_size"],
        weight_decay=config["weight_decay"],
        warmup_ratio=config["warmup_ratio"],
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="adult_f1",
        greater_is_better=True,
        save_total_limit=1,
        logging_strategy="epoch",
        report_to="none",
        fp16=torch.cuda.is_available(),
        seed=SEED,
    )
    trainer = make_trainer(
        model=model, args=args, train_dataset=train_ds, eval_dataset=val_ds,
        processing_class=tokenizer, data_collator=data_collator, compute_metrics=hf_metrics,
    )
    trainer.train()
    val_out = trainer.predict(val_ds)
    val_logits = val_out.predictions
    val_probs = torch.softmax(torch.tensor(val_logits), dim=-1).numpy()[:, 1]
    val_pred = np.argmax(val_logits, axis=-1)
    metrics = get_metrics(y_val, val_pred, val_probs)
    trial_records.append({"trial": trial, **config, **{f"val_{k}": v for k, v in metrics.items()}})
    print("Validation metrics:", metrics)

    if metrics["adult_f1"] > best_val_f1:
        best_val_f1 = metrics["adult_f1"]
        best_config = config.copy()
        best_trial_model_dir = best_dir
        best_dir.mkdir(parents=True, exist_ok=True)
        trainer.save_model(str(best_dir))
        tokenizer.save_pretrained(str(best_dir))
        print("Saved new best DistilBERT trial.")

    del trainer, model
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

trial_df = pd.DataFrame(trial_records)
trial_df.to_csv(OUTPUT_DIR / "distilbert_random_search_results.csv", index=False)
print("\nSearch results sorted by validation adult F1:")
print(trial_df.sort_values("val_adult_f1", ascending=False).to_string(index=False))
print("Best DistilBERT configuration:", best_config)

# ============================================================
# 7. EVALUATE BEST DISTILBERT AND SELECT OVERALL MODEL BY VALIDATION F1
# ============================================================
best_distilbert = AutoModelForSequenceClassification.from_pretrained(str(best_trial_model_dir))
eval_args = make_training_args(
    output_dir=str(TRIALS_DIR / "evaluation_only"),
    per_device_eval_batch_size=best_config["batch_size"],
    report_to="none",
    fp16=torch.cuda.is_available(),
)
eval_trainer = make_trainer(
    model=best_distilbert, args=eval_args, data_collator=data_collator,
    processing_class=tokenizer,
)
val_out = eval_trainer.predict(val_ds)
val_logits = val_out.predictions
val_prob = torch.softmax(torch.tensor(val_logits), dim=-1).numpy()[:, 1]
val_pred = np.argmax(val_logits, axis=-1)
distilbert_val_metrics = get_metrics(y_val, val_pred, val_prob)

# Test results for the selected DistilBERT configuration; do not use test results to select.
test_out = eval_trainer.predict(test_ds)
test_logits = test_out.predictions
test_prob = torch.softmax(torch.tensor(test_logits), dim=-1).numpy()[:, 1]
test_pred = np.argmax(test_logits, axis=-1)
all_validation = {**classical_validation, "DistilBERT (best randomized config)": distilbert_val_metrics}
all_test_outputs = {
    **classical_test_outputs,
    "DistilBERT (best randomized config)": {
        "y_pred": np.asarray(test_pred), "y_score": np.asarray(test_prob),
        "metrics": get_metrics(y_test, test_pred, test_prob),
    },
}
validation_df = pd.DataFrame(all_validation).T.sort_values("adult_f1", ascending=False)
validation_df.to_csv(OUTPUT_DIR / "validation_model_comparison.csv")

# Unified experiment log: both search procedures feed into one comparison file.
# Classical trials are scored with 3-fold CV during tuning; their selected
# estimator is then scored on the shared validation split. DistilBERT trials
# are ranked on that same validation split because full K-fold fine-tuning is costly.
classical_trial_rows = []
for model_name, search_result in classical_cv_results.items():
    classical_trial_rows.append({
        "model_family": "classical_ml",
        "model": model_name,
        "trial_type": "RandomizedSearchCV (3-fold CV)",
        "cv_adult_f1": search_result["best_cv_f1"],
        "validation_adult_f1": classical_validation[model_name]["adult_f1"],
        "best_parameters": json.dumps(search_result["best_params"], default=str),
    })
distilbert_trial_rows = []
for record in trial_records:
    distilbert_trial_rows.append({
        "model_family": "DistilBERT",
        "model": f"DistilBERT trial {record['trial']}",
        "trial_type": "Randomized configuration / validation split",
        "cv_adult_f1": np.nan,
        "validation_adult_f1": record.get("val_adult_f1", np.nan),
        "best_parameters": json.dumps({k: v for k, v in record.items() if k not in {"trial", "val_accuracy", "val_adult_precision", "val_adult_recall", "val_adult_f1", "val_roc_auc", "val_pr_auc"}}, default=str),
    })
unified_search_df = pd.DataFrame(classical_trial_rows + distilbert_trial_rows)
unified_search_df.to_csv(OUTPUT_DIR / "unified_search_results.csv", index=False)
print("\nUnified search log saved to:", OUTPUT_DIR / "unified_search_results.csv")
best_model_name = validation_df["adult_f1"].idxmax()
test_comparison_df = pd.DataFrame({name: output["metrics"] for name, output in all_test_outputs.items()}).T
test_comparison_df.to_csv(OUTPUT_DIR / "test_model_comparison.csv")
print("\nVALIDATION MODEL COMPARISON (used for model selection):")
print(validation_df.to_string())
print("\nSelected model based on validation adult F1:", best_model_name)
print("\nTEST METRICS (final reporting):")
print(test_comparison_df.to_string())

selected = all_test_outputs[best_model_name]
y_selected = selected["y_pred"]
scores_selected = selected["y_score"]
print("\nSelected model classification report:")
print(classification_report(y_test, y_selected, labels=[0, 1], target_names=["Non-Adult", "Adult"], zero_division=0))
print("Selected model confusion matrix:\n", confusion_matrix(y_test, y_selected, labels=[0, 1]))

# ============================================================
# 8. VISUALIZATION: CLASS DISTRIBUTION
# ============================================================
fig, ax = plt.subplots(figsize=(7, 5))
df["label"].map(ID2LABEL).value_counts().reindex(["non-adult", "adult"]).plot(kind="bar", ax=ax)
ax.set(title="Class Distribution After Preprocessing", xlabel="Class", ylabel="Samples")
ax.tick_params(axis="x", rotation=0)
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "class_distribution.png", dpi=150); plt.show()

# ============================================================
# 9. VISUALIZATION: DISTILBERT RANDOM SEARCH
# ============================================================
fig, ax = plt.subplots(figsize=(9, 5))
ordered_trials = trial_df.sort_values("val_adult_f1", ascending=False)
ax.bar([f"Trial {int(x)}" for x in ordered_trials["trial"]], ordered_trials["val_adult_f1"])
ax.set(title="DistilBERT Random Search: Validation Adult F1", xlabel="Trial", ylabel="Adult F1", ylim=(0, 1))
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "distilbert_random_search_f1.png", dpi=150); plt.show()

# Table visualization for the sampled hyperparameters and their validation metrics.
table_cols = [c for c in ["trial", "learning_rate", "epochs", "batch_size", "weight_decay", "warmup_ratio", "val_accuracy", "val_adult_precision", "val_adult_recall", "val_adult_f1", "val_roc_auc", "val_pr_auc"] if c in trial_df.columns]
fig, ax = plt.subplots(figsize=(13, max(3, 0.65 * len(trial_df))))
ax.axis("off")
tab = ax.table(cellText=trial_df[table_cols].round(4).values, colLabels=table_cols, loc="center", cellLoc="center")
tab.auto_set_font_size(False); tab.set_fontsize(8); tab.scale(1, 1.5)
ax.set_title("DistilBERT Hyperparameter Search Results", pad=18)
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "distilbert_hyperparameter_results.png", dpi=150); plt.show()

# ============================================================
# 10. VISUALIZATION: TEST METRIC MODEL COMPARISON
# ============================================================
plot_metrics = [c for c in ["accuracy", "adult_precision", "adult_recall", "adult_f1", "roc_auc", "pr_auc"] if c in test_comparison_df.columns]
fig, ax = plt.subplots(figsize=(12, 6))
test_comparison_df[plot_metrics].plot(kind="bar", ax=ax)
ax.set(title="Model Comparison on Held-Out Test Set", xlabel="Model", ylabel="Score", ylim=(0, 1))
ax.tick_params(axis="x", rotation=20)
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "test_model_comparison.png", dpi=150); plt.show()

# ============================================================
# 11. VISUALIZATION: ROC CURVES AND AUC
# ============================================================
fig, ax = plt.subplots(figsize=(8, 6))
for name, output in all_test_outputs.items():
    fpr, tpr, _ = roc_curve(y_test, output["y_score"])
    auc_value = roc_auc_score(y_test, output["y_score"])
    ax.plot(fpr, tpr, label=f"{name} (AUC={auc_value:.3f})")
ax.plot([0, 1], [0, 1], linestyle="--", label="Random baseline")
ax.set(title="ROC Curves on Held-Out Test Set", xlabel="False Positive Rate", ylabel="True Positive Rate")
ax.legend(loc="lower right")
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "roc_curves.png", dpi=150); plt.show()

# ============================================================
# 12. VISUALIZATION: PRECISION-RECALL CURVES
# ============================================================
fig, ax = plt.subplots(figsize=(8, 6))
for name, output in all_test_outputs.items():
    p, r, _ = precision_recall_curve(y_test, output["y_score"])
    ap = average_precision_score(y_test, output["y_score"])
    ax.plot(r, p, label=f"{name} (AP={ap:.3f})")
ax.set(title="Precision-Recall Curves on Held-Out Test Set", xlabel="Recall", ylabel="Precision")
ax.legend(loc="best")
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "precision_recall_curves.png", dpi=150); plt.show()

# ============================================================
# 13. VISUALIZATION: SELECTED MODEL CONFUSION MATRICES
# ============================================================
cm = confusion_matrix(y_test, y_selected, labels=[0, 1])
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm, interpolation="nearest")
ax.set(title=f"Confusion Matrix: {best_model_name}", xlabel="Predicted label", ylabel="True label")
ax.set_xticks([0, 1], labels=["Non-Adult", "Adult"]); ax.set_yticks([0, 1], labels=["Non-Adult", "Adult"])
for i in range(2):
    for j in range(2):
        ax.text(j, i, str(cm[i, j]), ha="center", va="center", color="white" if cm[i, j] > cm.max()/2 else "black")
fig.colorbar(im, ax=ax); fig.tight_layout()
fig.savefig(OUTPUT_DIR / "selected_confusion_matrix.png", dpi=150); plt.show()

row_totals = cm.sum(axis=1, keepdims=True)
cm_norm = np.divide(cm.astype(float), row_totals, out=np.zeros_like(cm, dtype=float), where=row_totals != 0)
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm_norm, interpolation="nearest", vmin=0, vmax=1)
ax.set(title=f"Normalized Confusion Matrix: {best_model_name}", xlabel="Predicted label", ylabel="True label")
ax.set_xticks([0, 1], labels=["Non-Adult", "Adult"]); ax.set_yticks([0, 1], labels=["Non-Adult", "Adult"])
for i in range(2):
    for j in range(2):
        ax.text(j, i, f"{cm_norm[i, j]:.2f}", ha="center", va="center", color="white" if cm_norm[i, j] > 0.5 else "black")
fig.colorbar(im, ax=ax); fig.tight_layout()
fig.savefig(OUTPUT_DIR / "selected_normalized_confusion_matrix.png", dpi=150); plt.show()

# ============================================================
# 14. VISUALIZATION: PER-CLASS METRICS
# ============================================================
report = classification_report(y_test, y_selected, labels=[0, 1], target_names=["Non-Adult", "Adult"], output_dict=True, zero_division=0)
per_class = pd.DataFrame(report).T.loc[["Non-Adult", "Adult"], ["precision", "recall", "f1-score"]]
fig, ax = plt.subplots(figsize=(8, 5))
per_class.plot(kind="bar", ax=ax)
ax.set(title=f"Per-Class Metrics: {best_model_name}", xlabel="Class", ylabel="Score", ylim=(0, 1))
ax.tick_params(axis="x", rotation=0)
fig.tight_layout(); fig.savefig(OUTPUT_DIR / "per_class_metrics.png", dpi=150); plt.show()

# ============================================================
# 15. SAVE MODEL ARTIFACTS
# ============================================================
# Save fitted classical pipelines as complete Joblib .pkl files.
for name, pipeline in classical_fitted.items():
    safe_name = name.lower().replace(" + ", "_").replace(" ", "_").replace("-", "_")
    joblib.dump(pipeline, MODEL_PATH / f"{safe_name}.pkl")

# Save actual DistilBERT weights/tokenizer in Hugging Face format.
distilbert_final_path = MODEL_PATH / "distilbert_best"
distilbert_final_path.mkdir(parents=True, exist_ok=True)
best_distilbert.save_pretrained(str(distilbert_final_path))
tokenizer.save_pretrained(str(distilbert_final_path))

metadata = {
    "selected_model_name": best_model_name,
    "selection_metric": "validation adult_f1",
    "validation_metrics": all_validation[best_model_name],
    "test_metrics": all_test_outputs[best_model_name]["metrics"],
    "label2id": LABEL2ID,
    "id2label": ID2LABEL,
    "max_length": MAX_LENGTH,
    "dataset_path": str(DATASET_PATH.resolve()),
    "distilbert_model_path": str(distilbert_final_path.resolve()),
    "best_distilbert_config": best_config,
    "classical_ml_cv_results": classical_cv_results,
    "note": "DistilBERT weights are saved in Hugging Face format. The metadata .pkl does not contain DistilBERT weights. Classical pipeline .pkl files contain fitted sklearn models."
}
joblib.dump(metadata, MODEL_PATH / "best_model_metadata.pkl")
with open(OUTPUT_DIR / "experiment_summary.json", "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=2, default=float)

print("\n" + "=" * 60 + "\nPIPELINE COMPLETED\n" + "=" * 60)
print("Selected model (based on validation adult F1):", best_model_name)
print("Selected model test metrics:", all_test_outputs[best_model_name]["metrics"])
print("Model artifacts:", MODEL_PATH.resolve())
print("Results and graphs:", OUTPUT_DIR.resolve())
print("\nGenerated graph files:")
for name in [
    "class_distribution.png", "distilbert_random_search_f1.png",
    "distilbert_hyperparameter_results.png", "test_model_comparison.png",
    "roc_curves.png", "precision_recall_curves.png",
    "selected_confusion_matrix.png", "selected_normalized_confusion_matrix.png",
    "per_class_metrics.png"
]:
    print(" -", OUTPUT_DIR / name)
```
